# Matplotlib

[Matplotlib](https://matplotlib.org/) is a comprehensive library for creating static, animated, and interactive visualizations. Install the Matplotlib Package using the command:

```bash
pip install matplotlib

```
Import the $\texttt{pyplot}$ module from Matplotlib using the conventional alias $\texttt{plt}$. 

### 1 3D Visualization
Matplotlib can be used to visualize three-dimensional data such as point clouds, trajectories, surfaces, and geometric structures.

In the previous NumPy exercises, we represented 3D points using an $N \times 3$ array:

$[[x1,y1,z1],\\
  [x2,y2,z2],\\
...\\
  [xn,yn,zn]]$
